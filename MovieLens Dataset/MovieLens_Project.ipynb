# 🎬 MovieLens Data Analysis Project
تحليل تقييمات الأفلام باستخدام مجموعة بيانات MovieLens الصغيرة.

```
# 📦 Step 1: Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set(style="darkgrid")
```

```
# 📂 Step 2: Load the dataset
ratings = pd.read_csv("ratings.csv")
movies = pd.read_csv("movies.csv")
```

```
# 🧹 Step 3: Merge datasets on movieId
data = pd.merge(ratings, movies, on="movieId")
```

```
# 🕵️ Step 4: Basic exploration
print("\nFirst 5 rows:")
print(data.head())
print("\nData Info:")
print(data.info())
print("\nMissing values:")
print(data.isnull().sum())
```

```
# 📊 Step 5: Most rated movies
most_rated = data.groupby('title').size().sort_values(ascending=False).head(10)
plt.figure(figsize=(10,5))
most_rated.plot(kind='bar', title='Top 10 Most Rated Movies')
plt.ylabel("Number of Ratings")
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```
# 📊 Step 6: Highest average ratings (with filter)
movie_stats = data.groupby('title')['rating'].agg(['count','mean'])
high_rated = movie_stats[movie_stats['count'] >= 100].sort_values(by='mean', ascending=False).head(10)
plt.figure(figsize=(10,5))
high_rated['mean'].plot(kind='bar', title='Top 10 Highest Rated Movies (min 100 ratings)')
plt.ylabel("Average Rating")
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```
# 🔍 Step 7: Ratings distribution
plt.figure(figsize=(8,5))
sns.histplot(data['rating'], bins=10, kde=True)
plt.title("Ratings Distribution")
plt.xlabel("Rating")
plt.ylabel("Count")
plt.show()
```

```
# 🔥 Step 8: Heatmap of correlation (can be slow for large datasets)
pivot_table = data.pivot_table(index='userId', columns='movieId', values='rating')
correlation_matrix = pivot_table.corr(method='pearson', min_periods=50)
sns.heatmap(correlation_matrix.fillna(0), cmap="coolwarm")
plt.title("Movies Correlation Heatmap")
plt.show()
```
